```python
from ultralytics import YOLO

# Load model nhỏ, nhanh
model = YOLO('yolov8n.pt')  # Nếu muốn chính xác hơn, thay bằng 'yolov8s.pt'

# Train
results = model.train(
    data='/content/Vietnamese-Food-1/data.yaml',
    epochs=60,              # 60 epochs đủ tốt cho dataset nhỏ
    imgsz=640,
    batch=16,               # Nếu lỗi RAM thì giảm xuống 8
    name='viet_food_5classes',
    patience=15,            # Dừng sớm nếu không cải thiện
    device=0                # GPU
)
```

```python
# Load model tốt nhất
model = YOLO('/content/runs/detect/viet_food_5classes/weights/best.pt')

# Export sang ONNX
model.export(format='onnx', opset=19)
```
